```python
import pandas as pd
```

```python
try:
    df = pd.read_csv('data.csv')
    print('CSV loaded successfully.')
    display(df.head())
except FileNotFoundError:
    print("Error: 'data.csv' not found. Please upload the CSV file or provide the correct path.")
except Exception as e:
    print(f"An error occurred: {e}")
```

```python
import numpy as np
from sklearn.preprocessing import MinMaxScaler

def create_sequences(data, seq_length, target_idx=0):
    X, y = [], []
    for i in range(len(data) - seq_length):
        X.append(data[i:i+seq_length])
        y.append(data[i+seq_length, target_idx])  # predict temperature
    return np.array(X), np.array(y)

# Scale the DataFrame to create scaled_data
scaler = MinMaxScaler()
scaled_data = scaler.fit_transform(df)

SEQ_LENGTH = 30  # past 30 days
X, y = create_sequences(scaled_data, SEQ_LENGTH)
```

```python
train_size = int(0.7 * len(X))
val_size = int(0.15 * len(X))

X_train, y_train = X[:train_size], y[:train_size]
X_val, y_val = X[train_size:train_size+val_size], y[train_size:train_size+val_size]
X_test, y_test = X[train_size+val_size:], y[train_size+val_size:]
```

```python
from tensorflow.keras.models import Sequential
from tensorflow.keras.layers import LSTM, Dense, Dropout
from tensorflow.keras.callbacks import EarlyStopping, ReduceLROnPlateau

model = Sequential([
    LSTM(64, return_sequences=True, input_shape=(SEQ_LENGTH, X.shape[2])),
    Dropout(0.2),
    LSTM(32, return_sequences=False),
    Dropout(0.2),
    Dense(16, activation='relu'),
    Dense(1)  # Predict single temperature value
])

model.compile(optimizer='adam', loss='mae', metrics=['mae', 'mse'])
```

```python
callbacks = [
    EarlyStopping(patience=10, restore_best_weights=True, monitor='val_loss'),
    ReduceLROnPlateau(factor=0.5, patience=5, min_lr=1e-6)
]

history = model.fit(
    X_train, y_train,
    validation_data=(X_val, y_val),
    epochs=100,
    batch_size=32,
    callbacks=callbacks,
    verbose=1
)
```

```python
y_pred = model.predict(X_test)
# Inverse scale temperature only
y_pred_orig = scaler.inverse_transform(np.hstack((y_pred, np.zeros((len(y_pred), X.shape[2]-1)))))[:, 0]
y_test_orig = scaler.inverse_transform(np.hstack((y_test.reshape(-1,1), np.zeros((len(y_test), X.shape[2]-1)))))[:, 0]
```

```python
from sklearn.metrics import mean_absolute_error, mean_squared_error, r2_score
mae = mean_absolute_error(y_test_orig, y_pred_orig)
rmse = np.sqrt(mean_squared_error(y_test_orig, y_pred_orig))
r2 = r2_score(y_test_orig, y_pred_orig)
print(f"MAE: {mae:.2f}°C | RMSE: {rmse:.2f}°C | R²: {r2:.3f}")
```

```python
model.save('lstm_temp_predictor.keras')
import joblib
joblib.dump(scaler, 'scaler.pkl')
```

```python
from fastapi import FastAPI
import numpy as np, joblib, tensorflow as tf

app = FastAPI()
model = tf.keras.models.load_model('lstm_temp_predictor.keras')
scaler = joblib.load('scaler.pkl')

@app.post("/predict")
def predict(payload: dict):
    # payload: {"features": [[t-29...t0, feat1, feat2...], ...]}
    x = np.array(payload["features"]).reshape(1, SEQ_LENGTH, -1)
    x_scaled = scaler.transform(x.reshape(-1, x.shape[2])).reshape(x.shape)
    pred = model.predict(x_scaled)
    pred_orig = scaler.inverse_transform(np.hstack((pred, np.zeros((1, x.shape[2]-1)))))[0, 0]
    return {"predicted_temperature": round(pred_orig, 2)}
```
